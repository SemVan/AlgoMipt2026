```python
#[1,7,15,64,72,87,91] 87

def binary_search(arr, num):
    if len(arr) == 0:
        return
    mid = len(arr) // 2
    if arr[mid] == num:
        return mid
    elif mid == 0:
        return 
    elif arr[mid] > num:
        return binary_search(arr[:mid], num)
    else:
        res = binary_search(arr[mid+1:], num)
        if res is not None:
            return res + mid + 1
        else:
            return res
```

```python
def binary_search_not_rec(arr, num):
    left = 0
    right = len(arr) - 1
    cnt = 0
    while left <= right:
        cnt += 1
        mid = (left + right) // 2
        if arr[mid] == num:
            return mid, cnt
        elif arr[mid] < num:
            left = mid + 1
        else:
            right = mid - 1
    return None, cnt
```

```python
binary_search([1], 1)
```

```python
binary_search_not_rec([1,7,15,64,72,87,91, 191], 191)
```
