```python
#[1,7,15,64,72,87,91] 87

def binary_search(arr, num):
    if len(arr) == 0:
        return
    mid = len(arr) // 2
    if arr[mid] == num:
        return mid
    elif mid == 0:
        return 
    elif arr[mid] > num:
        return binary_search(arr[:mid], num)
    else:
        res = binary_search(arr[mid+1:], num)
        if res is not None:
            return res + mid + 1
        else:
            return res
```

```python
def binary_search_not_rec(arr, num):
    left = 0
    right = len(arr) - 1
    cnt = 0
    while left <= right:
        cnt += 1
        mid = (left + right) // 2
        if arr[mid] == num:
            return mid, cnt
        elif arr[mid] < num:
            left = mid + 1
        else:
            right = mid - 1
    return None, cnt
```

```python
binary_search_not_rec([1,7,15,64,72,87,91, 191], 191)
```

```python
%time
import numpy as np
import time

cnts = []
lens = [100, 1000, 5000, 10000, 100000, 1000000]
times = []
for i in lens:
  cnts_loc = []
  times_loc = []
  for j in range(100):
    start = time.time()
    answer, cnt = binary_search_not_rec(sorted(np.random.randint(0, i, i)), np.random.randint(0, i))
    diff = time.time() - start
    times_loc.append(diff)
    cnts_loc.append(cnt)
  cnts.append(np.mean(cnts_loc))
  times.append(np.mean(times_loc))
```

```python
from matplotlib import pyplot as plt


plt.plot(lens, cnts, label='real')
plt.plot(lens, 1.4*np.log(lens), label='ideal')
# plt.plot(lens, lens, label='linear')
# plt.xlim(0, 10000)
```
