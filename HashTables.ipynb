```python
class HashTable:
    def __init__(self, size=1000):
        self.size = size
        self.table = [None] * size
        self.collision_count = 0

    def _get_index(self, key):
        return hash(key) % self.size

    def add(self, key, value):
        index = self._get_index(key)
        if self.table[index] is None:
            self.table[index] = value
        else:
            self.collision_count += 1

    def get(self, key):
        index = self._get_index(key)
        return self.table[index]

    def get_free_cells(self):
        cnt = 0
        for i in range(self.size):
            if self.table[i] is None:
                cnt += 1
        return cnt

        
```

```python
class Node:
    def __init__(self, key, value):
        self.key = key
        self.value = value
        self.next = None


class HashTableWithChains:
    def __init__(self, size=1000):
        self.size = size
        self.table = [None] * size
        self.collision_count = 0
        self.max_list_len = 0

    def _get_index(self, key):
        return hash(key) % self.size

    def _add_to_tail(self, node, new_node):
        cnt = 1
        while node.next:
            cnt += 1
            node = node.next
        if cnt > self.max_list_len:
            self.max_list_len = cnt
        node.next = new_node

    def _get_from_list(self, node, key):
        while node and node.key != key:
            node = node.next
        if node.key == key:
            return node.value
        return None

    def add(self, key, value):
        index = self._get_index(key)
        if self.table[index] is None:
            self.table[index] = Node(key, value)
        else:
            self.collision_count += 1
            self._add_to_tail(self.table[index], Node(key, value))
             

    def get(self, key):
        index = self._get_index(key)
        return self._get_from_list(self.table[index], key)
        

    def get_free_cells(self):
        cnt = 0
        for i in range(self.size):
            if self.table[i] is None:
                cnt += 1
        return cnt

        
```

```python
tab = HashTableWithChains(size=1)
for i in range(1000):
    tab.add(str(i), i)

for i in range(1000):
    res = tab.get(str(i))
    if res != i:
        print('failed')
```

```python
class HashTableOpenAddress:
    def __init__(self, size=1000):
        self.size = size
        self.table = [None] * size
        self.collision_count = 0

    def _get_index(self, key):
        return hash(key) % self.size

    def add(self, key, value):
        index = self._get_index(key)
        if self.table[index] is None:
            self.table[index] = (key, value)
        else:
            self.collision_count += 1
            step = 1
            while index+step < self.size or index-step >= 0:
                if index+step < self.size and self.table[index + step] is None:
                    self.table[index + step] = (key, value)
                    break
                if index-step >= 0 and self.table[index - step] is None:
                    self.table[index - step] = (key, value)
                    break
                step += 1
            if index+step >= self.size and index-step < 0:
                self._change_size()

    def _change_size(self):
        print('changing size')
        self.size = self.size * 2
        old_table = self.table
        self.table = [None] * self.size
        self.collision_count = 0
        for key, value in old_table:
            self.add(key, value)
        return

    def get(self, key):
        index = self._get_index(key)
        if self.table[index] is not None:
            if self.table[index][0] == key:
                return self.table[index][1]
            else:
                step = 1
                while index+step < self.size or index-step >= 0:
                    if index+step < self.size and self.table[index + step] is not None:
                        if self.table[index + step][0] == key:
                            return self.table[index + step][1]
                    elif index-step >= 0 and self.table[index - step] is not None:
                        if self.table[index - step][0] == key:
                            return self.table[index - step][1]
                    elif self.table[index + step] is None or self.table[index - step] is None:
                        return None
                    step += 1
                return None

    def get_free_cells(self):
        cnt = 0
        for i in range(self.size):
            if self.table[i] is None:
                cnt += 1
        return cnt

        
```

```python
tab = HashTableOpenAddress(size=1000)
for i in range(1000):
    tab.add(str(i), i)

for i in range(1000):
    res = tab.get(str(i))
    if i != res:
        print(i, res)
        print('failed')
```

```python
# Сравниваем ТОЛЬКО поиск свободной ячейки при вставке уникальных ключей.
# Основа — наш HashTableOpenAddress. Хранилище и начальный индекс те же;
# два варианта отличаются только генератором индексов при коллизии.
# Расширение и get() здесь не используются: они не входят в эксперимент.
# Алгоритм CPython: https://github.com/python/cpython/blob/3.14/Objects/dictobject.c

import random
import sys
from statistics import mean


class HashTableProbeCount(HashTableOpenAddress):
    def __init__(self, size):
        if size < 2 or size & (size - 1):
            raise ValueError('Для алгоритма Питерса size должен быть степенью двойки')
        super().__init__(size)
        self.used = 0
        self.probe_counts = []

    def _probe_indices(self, key):
        # Точно наш порядок: h, h+1, h-1, h+2, h-2, ... без выхода за границы.
        index = self._get_index(key)
        yield index
        step = 1
        while index + step < self.size or index - step >= 0:
            if index + step < self.size:
                yield index + step
            if index - step >= 0:
                yield index - step
            step += 1

    def _find_free_cell(self, key):
        # Один шаг = одно чтение ячейки, включая самую первую проверку.
        for probes, index in enumerate(self._probe_indices(key), start=1):
            if self.table[index] is None:
                return index, probes
        raise RuntimeError('Нет свободных ячеек')

    def add(self, key, value):
        # В эксперименте ключи уникальны: обновление существующего ключа
        # не требуется. Одинаковая вставка для обоих вариантов.
        if self.used == self.size:
            raise RuntimeError('Таблица заполнена; расширение исключено из сравнения')
        index, probes = self._find_free_cell(key)
        self.table[index] = (key, value)
        self.used += 1
        self.probe_counts.append(probes)
        self.collision_count += probes > 1


class HashTablePeters(HashTableProbeCount):
    def _probe_indices(self, key):
        mask = self.size - 1
        # В CPython perturb имеет беззнаковый тип размера Py_hash_t.
        # Это важно для отрицательных хешей: сдвиг должен дополнять нулями.
        hash_mask = (1 << sys.hash_info.width) - 1
        perturb = hash(key) & hash_mask
        index = perturb & mask  # то же, что hash(key) % size при size = 2**k
        while True:
            yield index
            perturb >>= 5
            index = (5 * index + 1 + perturb) & mask


def compare_probes(size=1024, trials=5, seed=42):
    loads = (0.50, 0.66, 0.85, 0.95)
    datasets = ('Случайные ключи', 'Одинаковые младшие биты', 'Последовательные ключи')
    print('Шаг = проверка одной ячейки при вставке нового ключа.')
    print(f'Размер: {size}; запусков на строку: {trials}.')
    print('Среднее — по всем вставкам; p95 — 95-й перцентиль числа шагов.')
    print('Выигрыш = среднее число шагов нашего алгоритма / Питерса; > 1 лучше у Питерса.')
    print('Это число обращений к таблице, а не замер времени Python-кода.\n')
    print(f'{"Набор":<29} {"Заполн.":>7} {"Наш ср.":>9} {"Питерс ср.":>11} '
          f'{"Наш p95":>8} {"Пит. p95":>9} {"Выигрыш":>9}')
    print('-' * 94)
    results = []
    for dataset in datasets:
        for load in loads:
            n = int(size * load)
            counts = [[], []]
            for trial in range(trials):
                rng = random.Random(seed + trial)
                if dataset == 'Случайные ключи':
                    keys = rng.sample(range(-(1 << 60), 1 << 60), n)
                elif dataset == 'Одинаковые младшие биты':
                    # На этих положительных int hash(key) == key.
                    # Начальная ячейка у всех — 0, но полные хеши разные.
                    keys = [i << (size.bit_length() + 5) for i in range(n)]
                    rng.shuffle(keys)
                else:
                    # Контроль: начальные индексы разные, коллизий нет.
                    keys = list(range(n))
                    rng.shuffle(keys)
                assert len(set(keys)) == n
                for j, cls in enumerate((HashTableProbeCount, HashTablePeters)):
                    table = cls(size)
                    for key in keys:
                        table.add(key, key)
                    # Проверяем сохранность всех пар без унаследованного get().
                    assert dict(pair for pair in table.table if pair is not None) == {k: k for k in keys}
                    counts[j].extend(table.probe_counts)
            averages = [mean(c) for c in counts]
            p95 = [sorted(c)[(95 * len(c) + 99) // 100 - 1] for c in counts]
            ratio = averages[0] / averages[1]
            print(f'{dataset:<29} {n / size:>7.1%} {averages[0]:>9.2f} '
                  f'{averages[1]:>11.2f} {p95[0]:>8} {p95[1]:>9} {ratio:>8.2f}x')
            results.append(dict(dataset=dataset, load=n / size,
                                ours_mean=averages[0], peters_mean=averages[1],
                                ours_p95=p95[0], peters_p95=p95[1], ratio=ratio))
    print('\nПри одинаковых младших битах perturb использует старшие биты хеша,')
    print('поэтому ключи расходятся по разным маршрутам вместо одной цепочки.')
    print('На последовательных ключах оба алгоритма делают ровно одну проверку.')
    print('Универсальной гарантии меньшего числа шагов у Питерса нет.')
    print('Заполненность 85–95% — стресс-тест: CPython обычно расширяет dict около 2/3.')
    return results


probe_results = compare_probes()
```
